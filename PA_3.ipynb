```python
import pandas as pd

cars = pd.read_csv('cars.csv')
```

### A. POSITIONAL AND LABEL-BASED SLICING

```python
print("Shape of Data Frame:", cars.shape)
print("Column Names:",  cars.columns.tolist())
cars_6_to_10 = cars.iloc[5:10]
cars_6_to_10_subset = cars_6_to_10.loc[:, ['Model', 'mpg', 'cyl', 'hp', 'gear']]
cars_6_to_10_subset
```

### B. MODEL LOOKUP

```python
toyota = cars[cars['Model'] == 'Toyota Corolla']
toyota
```

```python
pontiac = cars.loc[cars['Model'] == 'Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]
pontiac
```

### MULTI-MODEL SUBSETTING

```python
models_to_extract = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']
columns_to_keep = ['Model', 'mpg', 'cyl', 'hp', 'gear']
selected_cars = cars.loc[cars['Model'].isin(models_to_extract), columns_to_keep]
display(selected_cars)
print("Shape of selected_cars:", selected_cars.shape)
```
